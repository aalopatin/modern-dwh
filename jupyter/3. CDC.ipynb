## Function Write Changes

````python
def last_events(events_df, keys: list[str]):    
        
    columns_to_agg = [column for column in events_df.columns if column not in keys + ["__ts_ms"]]

    return (
        events_df
            .groupBy(keys)
            .agg(
                *[f.max_by(f.col(column), f.col("__ts_ms")).alias(column) for column in columns_to_agg]
            )
    )


def merge_events(target_table, events_df, keys: list[str]):

    keys_quoted = [f'`{k}`' for k in keys]
    
    on_condition = ' and '.join([f't.{k} = s.{k}' for k in keys_quoted])
    
    columns_to_insert_quoted = [f'`{c}`' for c in events_df.columns if c != "__op"]    
    insert_columns = ', '.join(columns_to_insert_quoted)
    insert_values  = ', '.join([f's.{c}' for c in columns_to_insert_quoted])
    update_columns = ', '.join([f't.{c} = s.{c}' for c in columns_to_insert_quoted if c not in keys])

    events_df.createOrReplaceTempView("events_df")
    
    events_df.sparkSession.sql(
        f'''
            MERGE INTO {target_table} t
            USING (SELECT * FROM events_df) s
            ON {on_condition}            
            WHEN MATCHED AND s.__op in ('r', 'c', 'u') THEN 
                UPDATE SET {update_columns}
            WHEN MATCHED AND s.__op = 'd' THEN 
                DELETE
            WHEN NOT MATCHED AND s.__op in ('r', 'c', 'u') THEN
                INSERT ({insert_columns}) VALUES ({insert_values})
        '''
    )


def write_changes(target_table, events_df, keys: str | list[str]):
    
    if isinstance(keys, str):
        keys = [keys]

    merge_events(
        target_table,
        events_df.transform(last_events, keys),        
        keys
    )
````

## Update Devices
###### silver.batch.operational_data.devices

````python
%%sparksql
CREATE TABLE silver.batch.operational_data.devices ( 
    device_id BIGINT NOT NULL, 
    device_code STRING NOT NULL, 
    model STRING, 
    description STRING
) USING iceberg 
````

````python
devices_table_name = 'silver.batch.operational_data.devices'

print('Devices before:')
spark.table(devices_table_name).show(truncate=False)

events_df = spark.table("bronze.cdc.operational_data.devices").where(f.col("__ts_ms") > "2025-09-17 00:00:00")

print('Events:')
events_df.show(truncate=False)

write_changes(
    devices_table_name,    
    events_df,
    'device_id'
)

print('Devices after:')
spark.table(devices_table_name).show(truncate=False)
````

````python
def write_devices(events_df, epoch_id):
    write_changes(
        'silver.stream.operational_data.devices',    
        events_df,
        'device_id'
    )
````

````python
from pyspark.sql.avro.functions import from_avro
from confluent_kafka.schema_registry import SchemaRegistryClient

kafka_boostrap = "kafka:9092"
schema_registy_url = "schema-registry:8081"
devices_topic_cdc = "cdc_flatten.operational_data.public.devices"

sr = SchemaRegistryClient({
    'url': f'http://{schema_registy_url}'
})

value_avro_schema = sr.get_latest_version(f'{devices_topic_cdc}-value').schema.schema_str

stream_df = (
    spark
        .readStream
        .format("kafka")
        .option("kafka.bootstrap.servers", kafka_boostrap)
        .option("subscribe", devices_topic_cdc)
        .option("startingOffsets", "earliest")
        .load()
)

streaming_query = (
    stream_df
        .selectExpr("substring(value, 6) as value")
        .select(
            from_avro("value", value_avro_schema).alias("value")
        )
        .select("value.*")
        .writeStream
        .option("checkpointLocation", "s3a://checkpoints/streaming/cdc/devices")
        .foreachBatch(
            lambda events_df, epoch_id: write_changes(
                'silver.stream.operational_data.devices',
                events_df,
                'device_id'
            )
        )
        .start()
)
````

````python
streaming_query.stop()
````
